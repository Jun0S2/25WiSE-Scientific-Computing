### 2. Sort the sequence of functions below: (i) assuming asymptotic growth, where g ∈ O(f) holds true for the functions g, f in a sorted sequence, and (ii) for n = 74 and give a short explanation.

* $4 \sqrt n $
* $ 2n^3 + 4n^2 + 950$
* $ \frac{n}{10000}\log_2{n} $
* $  \frac {1} {100} n! $
* $ 45 \log_2{n^2} + 9$
* $1$
* $ 10^{110}n +110 $
* $ 10n +3^n$

### (i) Identify function type
* $4 \sqrt n $ : root/polynomial
* $ 2n^3 + 4n^2 + 950$ : polynomial
* $ \frac{n}{10000}\log_2{n} $ : n log n
* $  \frac {1} {100} n! $ : factorial
* $ 45 \log_2{n^2} + 9$ : logarithmic
* $1$ : constant
* $ 10^{110}n +110 $ : linear
* $ 10n +3^n$ : exponential

### (ii) Standard asymptotic hierachy

For large n, ignore constants :

$$
1 << log n <<  n^a << c ^ n << n!
$$
for any c

so asymptoticall order is : 

$$
1  \lt \log n \lt \sqrt n \lt n\lt n\log n \lt n^3 \lt 3^n \lt n!
$$

### (iii) n = 74
* $4 \sqrt n = 3.440930e + 01$ 
* $ 2n^3 + 4n^2 + 950 = 8.8333020e_05$
* $ \frac{n}{10000}\log_2{n} = 4.594995e-02$
* $  \frac {1} {100} n! = 3.307885e+105$
* $ 45 \log_2{n^2} + 9 = 5.678508e+02$ 
* $1 = 1$

### (iv) Order comparison
Assymptotically: 
$$
1  \lt 45 \log_2 (n^2)+9 \lt 4\sqrt n \lt 10^{110}n+110 \lt  \frac n{100000}\log_2 n \lt 2*n^3+4*n^2+950\lt 10n + 3^n \lt \frac {1}{100}n!
$$

For n = 74, 
* $ \frac n {10000} log n $ is theoretically larger than constant 1, but due to its factor, its smaller for this n 
* Linear term $ 10^{110} n$ is asymptotically O(n), but the coefficient makes astronomically large for n=74

`answer`:

$$
 \frac n{100000}\log_2 n \lt 1  \lt 45 \log_2 (n^2)+9 \lt 4\sqrt n  \lt  2*n^3+4*n^2+950\lt 10n + 3^n \lt \frac {1}{100}n!\lt 10^{110}n+110
$$

### (v) short explanation
With above observation, asymptotic != numeric. asymptotic order ignores constant, while numerical comparison at a fixed n is dominated by them.
