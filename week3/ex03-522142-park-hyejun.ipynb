### 3. Find out which rounding mode Rust is using for floating point computations.

> Rust uses round to nearest, ties to even as its default rounding mode for floating-point arithmetic (f32 and f64).
> Reference : https://doc.rust-lang.org/reference/types/numeric.html 
> The types f32 and f64 correspond to the IEEE 754-2008 binary32 and binary64 floating-point formats.
> Because IEEE 754 specifies round to nearest, ties to even as the default rounding mode, Rust arithmetic follows this behavior by default.

### 4. If we have a 8-bit IEEE754 type floating point number (E4M3: sign, 4-bit exponent, 3-bit mantissa) which values can we represent?

> 1. E4M3 : 
> 
> | Field                   | Bits | Notes                                                          |
> | ----------------------- | ---- | -------------------------------------------------------------- |
> | Sign (S)                | 1    | 0 = positive, 1 = negative                                     |
> | Exponent (E)            | 4    | 4-bit exponent                                                 |
> | Mantissa / Fraction (M) | 3    | Fraction bits (with implicit leading 1 for normalized numbers) |
>
>
> reference:  https://en.wikipedia.org/wiki/Extended_precision
>
> 
> 2. Representable values in E4M3
>
> * **Normalized numbers** (E = 1..14):  $$ [\text{value} = (-1)^{\text{sign}} \times (1 + \text{mantissa}/8) \times 2^{E-7}] $$
> * **Subnormal numbers** (E = 0, mantissa ≠ 0): $$ \text{value} = (-1)^{\text{sign}} \times (\text{mantissa}/8) \times 2^{-6} $$
> * **Zero**: E = 0, mantissa = 0 → ±0
> * **Infinity**: E = 15, mantissa = 0 → ±∞
> * **NaN**: E = 15, mantissa ≠ 0 → NaN
>
> 3. Approximate numeric ranges
>
>  | Type               | Positive range    |
> | ------------------ | ----------------- |
> | Smallest subnormal | 1/512 ≈ 0.00195   |
> | Largest subnormal  | 7/512 ≈ 0.0137    |
> | Smallest normal    | 1 × 2^-6 ≈ 0.0156 |
> | Largest normal     | 1.875 × 2^7 ≈ 240 |
> | Infinity           | +∞                |
> 
> * Multiply all by −1 for negatives.
