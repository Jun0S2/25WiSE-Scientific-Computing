### 02. Let ` x:i32` , when  `x & (x-1) ` is true, why?


* If x is a power of 2, the result is 0 → false.
* If x is not a power of 2, the result is nonzero → true.

### Example
**Power of 2**

x = 8 = $ 1000_2 $
x-1 = 7 = $ 0111 $
> They are opposite so if you & operate this, this becomes 0 (false)

**Not power of 2 **

x = 2 = $ 10 _2 $
x-1 = 1 = $ 01_2 $
> After and operation, this becomes 0 (false)

If x is not a power of 2, then x has more than one 1-bit. In that case, at least one 1-bit remains after x & (x-1), so the result is always nonzero (true).



### 03. The following are binary values of some precision (32 bit) IEEE754 floating point values:

```
3FCF 1BBD
FF80 0000
8800 0000
3E4C CCCD
3FDD B3D7
```

### Which numbers do they represent exactly in decimal and which real number are they supposed to represent?

Example: C000 000 = exact decimal -2.0 real -2.0


---

### Approach
The IEEE 754 standard specifies a binary32 as having:


* Sign bit: 1 bit
* Exponent width: 8 bits
* Significand precision: 24 bits (23 explicitly stored) 

$$
N = (-1) ^s \times (1.f) \times 2^{e-127}
$$

* [IEEE Reference](https://en.wikipedia.org/wiki/Single-precision_floating-point_format)
* [Equation Reference](http://www.c-jump.com/bcc/common/Talk2/Cxx/IEEE_754_fp_standard/IEEE_754_fp_standard.html)


### Hex to 32bit Binary

Example : 3FCF1BBD -> 00111111 11001111 00011011 10111101

### Sign, Exponent, Fraction
* sign = 1 bit ( 0 - positive, 1 - negative )
* exponent = 8 bits
* fraction = last 23 bits

### Calculate Exponent
Exponent Field : Decimal - bias 127 (we need bias for the negative values)

### Calculate Mantissa 
> Mantissa = 1 + fraction(23 bits part)
Fraction = $ b_1 \times 2^{-1} + b_2 \times  2^{-2}...$

### Final Value
$$
N = (-1) ^s \times (1.f) \times 2^{e-127}
$$


--------------------------------------------------------------
### Hex: 3FCF1BBD
1. Binary: 00111111 11001111 00011011 10111101
2. Sign/Exponent/Fraction
* Sign bit: 0 (positive)
* Exponent field (raw 8 bits): 01111111 (decimal 127)
* Fraction bits (23 bits): 10011110001101110111101
4. Actual exponent (E = exponent_field - bias): 0
5. Mantissa = 1 + fraction (→ fraction = sum b_i*2^{-i} )
    Mantissa exact (as fraction) = 13573053/2^23 = 13573053 / 8388608

> Approximate decimal (f32, 17 significant digits) = 1.61803400516510010
